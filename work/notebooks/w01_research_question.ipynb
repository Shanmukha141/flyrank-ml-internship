## 2. The question: decision, action, cost of a wrong call

**Research question:** Which pages should be prioritised for content review based on observable search, content, and engagement signals?

**Unit of analysis:** One pseudonymised content page.

**Decision:** Decide which pages should be reviewed first when the SEO or content team has limited review capacity.

**Output:** A ranked list of pages with a priority score and understandable reason codes.

**Action:** A human reviewer can inspect the highest-ranked pages and decide whether to refresh, expand, protect, prune, or monitor them.

**Cost of a wrong call:** A false positive could waste reviewer time on a page that does not need attention. A false negative could cause the team to miss an important page whose performance is weakening or that represents a useful opportunity.

**Why data/ML can help:** Page performance involves multiple observable signals. Data analysis can identify useful patterns, and ML may combine these signals into a better ranking than a simple rule. I will only consider the added complexity useful if it improves on a transparent baseline under honest validation.

## 3. Quick look at the data (2-3 real numbers)

I used the anonymised starter dataset to check whether Refresh / Content Opportunity Scoring is worth investigating.

The following analysis looks at the size of the dataset, the proportion of pages currently labelled as declining, and the relationship between keyword search volume and actual page impressions.

These numbers help me check whether there is enough data and whether relying on a single intuitive signal such as search volume would be sufficient for prioritising pages.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Number of pages
print("Number of pages:", len(df))

# 2. Percentage currently labelled as declining
declining_rate = df["trend_direction"].str.lower().eq("down").mean() * 100
print(f"Declining pages: {declining_rate:.2f}%")

# 3. Search volume vs actual impressions
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Search volume vs impressions correlation: {corr:.3f}")
```
